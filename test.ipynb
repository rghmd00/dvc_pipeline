```python
import requests

data = {
    "columns":[
        "Pclass",
        "Sex",
        "Age",
        "SibSp",
        "Parch",
        "Fare",
        "Embarked"
    ],
    "rows":[
        [3,1,22.0,1,0,7.25,2]
    ] 
}
url = "http://localhost:8000/predict"

# Send the POST request
response = requests.post(url, json=data)

# Print the response
print("Response:", response.json())
if response.status_code == 200:
    predictions = response.json()
    print(predictions)
```
